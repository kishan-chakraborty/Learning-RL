**Implementing Monte-Carlo Method in Frozen-Lake environment.**

```python
import gymnasium as gym
```

```python
# Create the FrozenLake environment
env = gym.make(
    "FrozenLake-v1", render_mode="human"
)  # Use "rgb_array" for visual output

# Reset the environment
obs, info = env.reset()

# Run a random policy
for _ in range(100):
    action = env.action_space.sample()  # Take a random action
    obs, reward, terminated, truncated, info = env.step(action)

    if terminated or truncated:
        obs, info = env.reset()

env.close()
```
