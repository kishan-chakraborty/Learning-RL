```python
# Initialize the value and policy matrix.
gamma = 1
value = np.zeros(101)
policy = np.zeros(101, dtype=int)
reward = np.zeros(101)
reward[100] = 1


# Calculate expected reward for each state-action pair.
def expected_reward(state: int, action: int, p_h: float) -> float:
    global value
    """
    Expected reward for a state-action pair.
    """
    win = state + action
    loss = state - action

    bellman_expectation = p_h * (reward[win] + gamma * value[win]) + (1 - p_h) * (
        reward[loss] + gamma * value[loss]
    )
    return bellman_expectation


# Policy evaluation algorithm.
def policy_evaluation(eps=0.01, p_h=0.25):
    global value, policy
    """
    Find the value function corresponding to each state for a given policy.
    """
    while True:
        delta = 0
        for state in range(1, 100):
            v = value[state]
            value[state] = expected_reward(state, policy[state], p_h)
            delta = max(delta, abs(v - value[state]))

        if delta < eps:
            break

    return value


def policy_improvement(p_h=0.25):
    global policy
    """
    Improving the policy by greedy method.
    """
    policy_stable = True
    for state in range(1, 100):
        old_action = policy[state]

        max_action_val = -np.inf
        max_action = None

        # Iterate over all possible actions for the state state.
        for action in range(1, min(state, 100 - state) + 1):
            action_val = expected_reward(state, action, p_h)

            if action_val > max_action_val:
                max_action_val = action_val
                max_action = action

        policy[state] = max_action

        if old_action != max_action:
            policy_stable = False

    return policy_stable


# Implementing policyt iteration algorithm.
while True:
    policy_evaluation()
    policy_stable = policy_improvement()
    if policy_stable:
        break
```

```python
import numpy as np

# Parameters
gamma = 1.0  # Discount factor
p_h = 0.4  # Probability of heads
theta = 1e-5  # Convergence threshold

# Initialize value function and policy
value = np.zeros(101)  # Value function for states 0 to 100
policy = np.zeros(101, dtype=int)  # Policy for states 0 to 100
reward = np.zeros(101)  # Reward for each state
reward[100] = 1  # Reward for reaching state 100


def expected_reward(state: int, action: int, p_h: float) -> float:
    """
    Calculate the expected reward for a given state-action pair.
    """
    win = min(state + action, 100)  # Ensure win state does not exceed 100
    loss = max(state - action, 0)  # Ensure loss state does not go below 0

    # Bellman expectation equation
    bellman_expectation = p_h * (reward[win] + gamma * value[win]) + (1 - p_h) * (
        reward[loss] + gamma * value[loss]
    )
    return bellman_expectation


def value_iteration():
    """
    Perform value iteration to compute the optimal value function and policy.
    """
    global value, policy

    while True:
        delta = 0
        for state in range(1, 100):  # Exclude terminal states 0 and 100
            old_val = value[state]
            max_action_val = float("-inf")
            best_action = 0

            # Iterate over all possible actions
            for action in range(1, min(state, 100 - state) + 1):
                action_val = expected_reward(state, action, p_h)
                if action_val > max_action_val:
                    max_action_val = action_val
                    best_action = action

            # Update value function and policy
            value[state] = max_action_val
            policy[state] = best_action
            delta = max(delta, abs(old_val - value[state]))

        # Check for convergence
        if delta < theta:
            break


def display_results():
    """
    Display the optimal value function and policy.
    """
    print("Optimal Value Function:")
    print(value)
    print("\nOptimal Policy:")
    print(policy)


# Run value iteration
value_iteration()

# Display results
display_results()
```

```python
# Implementing policyt iteration algorithm.
while True:
    policy_stable = value_iteration()
    if policy_stable:
        break
```
