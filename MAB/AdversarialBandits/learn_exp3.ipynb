# Exp3 Algorithm
Problem: Adversarial Coin Flips
You are given 
N=10 actions, each corresponding to flipping a biased coin. <br> 
The probabilities of getting a "win" (heads) for each coin are defined as
$P_{1},P_{2},\cdots P_{10}$

Actions: Each action corresponds to selecting one of the 10 coins to flip.

Payoff:

If the coin lands heads, the reward is 1.
If the coin lands tails, the reward is 0.
Adversarial Setup:

At each round t, the adversary can manipulate the payoffs of the coins (i.e., the probabilities $P_{i}$) <br> 
to make the bandit algorithm perform poorly. The adversary ensures that the best coin changes dynamically, <br> 
or it designs the payoffs to mislead the algorithm during exploration and exploitation.<br>


Goal: Use the Exp3 algorithm to minimize regret, where regret is the difference between the cumulative <br> 
reward of the best coin (in hindsight) and the cumulative reward of your selected actions.
